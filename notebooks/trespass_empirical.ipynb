```python
import networkx as nx
# Load the route graph
G = nx.read_gml("../data/graph/LEMD_EGLL_2023_04_01.gml")
node_to_idx = {node: i for i, node in enumerate(G.nodes())}
idx_to_node = {i: node for i, node in enumerate(G.nodes())}
```

```python
flight_plan = 'LEMD ARLUN_82 BLV BISKA BADRU ECRIM GODEM_01 EGLL'
# Parse the flight plan and convert to node indices
waypoints = flight_plan.split()
waypoint_indices = [node_to_idx[waypoint] for waypoint in waypoints]

# Generate list of tuples representing consecutive waypoint pairs
waypoint_pairs = [(waypoint_indices[i], waypoint_indices[i+1]) for i in range(len(waypoint_indices)-1)]
print(f"Waypoint pairs: {waypoint_pairs}")
```

```python
# Create a tensor of shape (N, N) where N is the number of nodes in the graph,
# and the value at index (i, j) in waypoint_pairs is 1

import torch

# Get the number of nodes in the graph
num_nodes = len(G.nodes())

# Create a tensor of zeros with shape (num_nodes, num_nodes)
empirical_counts = torch.zeros((num_nodes, num_nodes), dtype=torch.float64)

# Set the value to 1 for each waypoint pair in the flight plan
for i, j in waypoint_pairs:
    empirical_counts[i, j] = 1.0

print(f"Empirical counts tensor shape: {empirical_counts.shape}")
print(f"Non-zero entries: {empirical_counts.nonzero().shape[0]}")
print(f"Waypoint pairs with counts:")
for i, j in waypoint_pairs:
    print(f"  {idx_to_node[i]} -> {idx_to_node[j]}: {empirical_counts[i, j].item()}")

# Save the empirical counts tensor to a file
torch.save(empirical_counts, "../data/empirical_counts/LEMD_EGLL_2023_04_01_empirical_counts.pt")
print(f"Empirical counts saved to ../data/empirical_counts/LEMD_EGLL_2023_04_01_empirical_counts.pt")
```
